```python

def prepare_data_for_clustering(df):
    """Prepare the data for clustering by separating features and scaling"""
   
    X = df.drop('Video_ID', axis=1).values
    image_names = df['Video_ID'].values
    
    
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    return X_scaled, image_names, scaler

X_scaled, image_names, scaler = prepare_data_for_clustering(aggregated_data)
```

```python

def find_optimal_clusters(X, max_k=10):
    """Determine the optimal number of clusters using silhouette score"""
    silhouette_scores = []
    
    for k in range(2, max_k + 1):
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(X)
        score = silhouette_score(X, kmeans.labels_)
        silhouette_scores.append(score)
        print(f"Silhouette score for k={k}: {score}")
    
  
    plt.figure(figsize=(10, 6))
    plt.plot(range(2, max_k + 1), silhouette_scores, 'bo-')
    plt.title('Silhouette Score Method')
    plt.xlabel('Number of clusters')
    plt.ylabel('Silhouette Score')
    plt.grid(True)
    plt.show()
    
  
    return silhouette_scores.index(max(silhouette_scores)) + 2


optimal_k = find_optimal_clusters(X_scaled)
print(f"Optimal number of clusters: {optimal_k}")
```

```python

def perform_clustering(X, k):
    """Perform KMeans clustering"""
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    return kmeans, labels

optimal_k = 3
kmeans, labels = perform_clustering(X_scaled, optimal_k)
```

```python
def visualize_clusters_pca_with_centroids(X, labels, image_names):
    """Visualize clusters in 2D using PCA with a side-by-side plot showing centroids"""

    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)
    
 
    pca_df = pd.DataFrame({'PCA1': X_pca[:, 0], 'PCA2': X_pca[:, 1], 
                          'Cluster': labels, 'Image': image_names})
    
 
    n_clusters = len(np.unique(labels))
    centroids_orig = np.zeros((n_clusters, X.shape[1]))
    
    for i in range(n_clusters):
        centroids_orig[i] = np.mean(X[labels == i], axis=0)
    
   
    centroids_pca = pca.transform(centroids_orig)
    
   
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))
    
   
    sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster', data=pca_df, palette='viridis', s=80, ax=ax1)
    ax1.set_title('All Data Points in PCA Space')
    ax1.legend(title='Cluster')
    

    sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster', data=pca_df, palette='viridis', 
                   s=30, alpha=0.2, legend=False, ax=ax2)
    

    for i in range(n_clusters):
        ax2.scatter(centroids_pca[i, 0], centroids_pca[i, 1], s=200, c=f'C{i}', 
                   marker='*', edgecolors='black', linewidth=1.5, label=f'Centroid {i}')
        ax2.annotate(f'Cluster {i}', (centroids_pca[i, 0], centroids_pca[i, 1]),
                   xytext=(10, 5), textcoords='offset points', fontsize=12, fontweight='bold')
    
    ax2.set_title('Cluster Centroids in PCA Space')
    ax2.legend(title='Centroids', loc='upper right')
    
   
    explained_var = pca.explained_variance_ratio_
    fig.suptitle(f'Cluster Visualization (PCA1: {explained_var[0]:.1%} variance, PCA2: {explained_var[1]:.1%} variance)',
                fontsize=16)
    
    plt.tight_layout()
    plt.show()
    
    return pca, X_pca, centroids_pca


pca, X_pca, centroids_pca = visualize_clusters_pca_with_centroids(X_scaled, labels, image_names)
```

```python

def get_distinctive_features(centers_df, top_n=5):
    distinctive_features = {}
    
    for i in range(centers_df.shape[0]):
       
        diff_from_mean = centers_df.iloc[i] - centers_df.mean()
       
        top_high = diff_from_mean.nlargest(top_n)
        
        top_low = diff_from_mean.nsmallest(top_n)
        
        distinctive_features[f'Cluster {i}'] = {
            'higher_than_avg': top_high.to_dict(),
            'lower_than_avg': top_low.to_dict()
        }
    
    return distinctive_features

centers = kmeans.cluster_centers_
au_columns = [col for col in aggregated_data.columns if col.startswith('AU')]
centers_df = pd.DataFrame(centers, columns=au_columns)
distinctive_features = get_distinctive_features(centers_df)


au_meaning = {
    'AU01': 'Inner Brow Raiser',
    'AU02': 'Outer Brow Raiser',
    'AU04': 'Brow Lowerer',
    'AU05': 'Upper Lid Raiser',
    'AU06': 'Cheek Raiser',
    'AU07': 'Lid Tightener',
    'AU09': 'Nose Wrinkler',
    'AU10': 'Upper Lip Raiser',
    'AU12': 'Lip Corner Puller (Smile)',
    'AU14': 'Dimpler',
    'AU15': 'Lip Corner Depressor',
    'AU17': 'Chin Raiser',
    'AU20': 'Lip Stretcher',
    'AU23': 'Lip Tightener',
    'AU25': 'Lips Part',
    'AU26': 'Jaw Drop',
    'AU45': 'Blink'
}


print("\n===== CLUSTER ANALYSIS =====")
for cluster, features in distinctive_features.items():
    print(f"\n{cluster} Distinctive Features:")
    
    print("  Higher than average intensity:")
    for feature, value in features['higher_than_avg'].items():
        au_code = feature.split('_')[0]  
        meaning = au_meaning.get(au_code, "Unknown")
        print(f"    {feature}: {value:.2f} - {meaning}")
    
    print("  Lower than average intensity:")
    for feature, value in features['lower_than_avg'].items():
        au_code = feature.split('_')[0]
        meaning = au_meaning.get(au_code, "Unknown")
        print(f"    {feature}: {value:.2f} - {meaning}")
```

```python

base_path= "C:/Users/alexa/Thesis/OpenFace_2.2.0_win_x64/processed"

    
def find_representative_images(df_clustered, X_scaled, kmeans, n=10, images_per_row=5):
    """Find and display the most representative images in each cluster (closest to centroid)"""
    representatives = {}
    
    
    n_clusters = len(kmeans.cluster_centers_)
    
   
    cluster_colors = plt.cm.tab10(np.linspace(0, 1, n_clusters))
    
  
    for cluster_id in range(n_clusters):
        
        in_cluster = df_clustered['cluster'] == cluster_id
        cluster_size = sum(in_cluster)
        
        if cluster_size == 0:
            continue
            
        
        cluster_points = X_scaled[in_cluster]
        centroid = kmeans.cluster_centers_[cluster_id].reshape(1, -1)
        distances = euclidean_distances(cluster_points, centroid)
        
        
        closest_indices = np.argsort(distances.flatten())[:n]
        
      
        original_indices = np.where(in_cluster)[0][closest_indices]
        representative_images = df_clustered.iloc[original_indices]
        
        representatives[cluster_id] = representative_images
        
       
        rows = int(np.ceil(min(n, cluster_size) / images_per_row))
        plt.figure(figsize=(images_per_row * 3, rows * 3 + 1))
        
      
        plt.suptitle(f"Cluster {cluster_id}: {cluster_size} images", 
                    fontsize=16, y=0.98)
        
       
        for i, (idx, img) in enumerate(representative_images.iterrows()):
            if i >= n:
                break
                
            ax = plt.subplot(rows, images_per_row, i + 1)
            
            
            ax.spines['bottom'].set_color(cluster_colors[cluster_id])
            ax.spines['top'].set_color(cluster_colors[cluster_id]) 
            ax.spines['right'].set_color(cluster_colors[cluster_id])
            ax.spines['left'].set_color(cluster_colors[cluster_id])
            ax.spines['bottom'].set_linewidth(5)
            ax.spines['top'].set_linewidth(5) 
            ax.spines['right'].set_linewidth(5)
            ax.spines['left'].set_linewidth(5)
            
          
            filename = img['Video_ID']
            short_name = filename[:15] + '...' if len(filename) > 15 else filename
            
         
            image_path = f"{base_path}/{filename}.jpg" 
            try:
                image = plt.imread(image_path)
                plt.imshow(image)
                plt.title(short_name, fontsize=10)
                plt.axis('off')
            except Exception as e:
                plt.text(0.5, 0.5, f"Missing image", ha='center', va='center')
                plt.gca().set_facecolor('#f0f0f0')  
                
        plt.tight_layout(rect=[0, 0, 1, 0.95])  
        plt.show()
    
    return representatives


representative_images = find_representative_images(aggregated_data, X_scaled, kmeans, n=200, images_per_row=6)
```

```python

emotion_mappings = {
    'Happiness': {
        'required': ['AU06', 'AU12'],
        'optional': ['AU25']
    },
    'Sadness': {
        'required': ['AU01', 'AU04', 'AU15'],
        'optional': ['AU17']
    },
    'Anger': {
        'required': ['AU04', 'AU07'],
        'optional': ['AU05', 'AU23', 'AU17']
    },
    'Fear': {
        'required': ['AU01', 'AU02', 'AU04'],
        'optional': ['AU05', 'AU20', 'AU26']
    },
    'Surprise': {
        'required': ['AU01', 'AU02', 'AU05'],
        'optional': ['AU26']
    },
    'Disgust': {
        'required': ['AU09', 'AU10'],
        'optional': ['AU15', 'AU17']
    },
    'Contempt': {
        'required': ['AU14'],
        'optional': []
    }
}


def identify_emotions_in_clusters(centers_df, threshold=0.5):
    """
    Identify likely emotions in each cluster based on AU patterns in centroids.
    threshold: minimum value to consider an AU as activated
    """
    cluster_emotions = {}
    
   
    
    for i in range(centers_df.shape[0]):
        cluster_emotions[f'Cluster {i}'] = []
        
        
        activated_aus = set()
        for col in centers_df.columns:
            if centers_df.iloc[i][col] > threshold:
                au_code = col.split('_')[0]
                activated_aus.add(au_code)
        
        
        for emotion, mapping in emotion_mappings.items():
            required_match = all(au in activated_aus for au in mapping['required'])
            
            if required_match and mapping['required']:  
                optional_count = sum(1 for au in mapping['optional'] if au in activated_aus)
                confidence = (len(mapping['required']) + optional_count) / (len(mapping['required']) + len(mapping['optional']))
                cluster_emotions[f'Cluster {i}'].append((emotion, confidence))
    
    return cluster_emotions


emotion_threshold = 0.5  
cluster_emotions = identify_emotions_in_clusters(centers_df, emotion_threshold)


print("\n===== CLUSTER EMOTION ANALYSIS =====")
for cluster, emotions in cluster_emotions.items():
    print(f"\n{cluster}:")
    if emotions:
        for emotion, confidence in sorted(emotions, key=lambda x: x[1], reverse=True):
            print(f"  {emotion} (confidence: {confidence:.2f})")
    else:
        print("  No clear emotional expressions identified")


print("\nCentroid value distribution:")
print(centers_df.describe())


for i in range(centers_df.shape[0]):
    print(f"\nCluster {i} value range: {centers_df.iloc[i].min():.2f} to {centers_df.iloc[i].max():.2f}")
    print(f"Number of AUs with value > 0.5: {(centers_df.iloc[i] > 0.5).sum()}")
    print(f"Number of AUs with value > 0.25: {(centers_df.iloc[i] > 0.25).sum()}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import numpy as np
import pandas as pd


au_meaning = {
    'AU01': 'Inner Brow Raiser',
    'AU02': 'Outer Brow Raiser',
    'AU04': 'Brow Lowerer',
    'AU05': 'Upper Lid Raiser',
    'AU06': 'Cheek Raiser',
    'AU07': 'Lid Tightener',
    'AU09': 'Nose Wrinkler',
    'AU10': 'Upper Lip Raiser',
    'AU12': 'Lip Corner Puller',
    'AU14': 'Dimpler',
    'AU15': 'Lip Corner Depressor',
    'AU17': 'Chin Raiser',
    'AU20': 'Lip Stretcher',
    'AU23': 'Lip Tightener',
    'AU25': 'Lips Part',
    'AU26': 'Jaw Drop',
    'AU45': 'Blink'
}

def analyze_cluster_performance(merged_data):
    """
    Comprehensive analysis of cluster performance with statistical tests
    """
    try:
     
        engagement_columns = ['View Count', 'Like Count', 'Comment Count']
        au_columns = [col for col in merged_data.columns if col.startswith('AU')]
        
       
        print("\n===== CLUSTER PERFORMANCE COMPARISON =====")
        performance_stats = merged_data.groupby('cluster')[engagement_columns].agg([
            'mean', 'median', 'std', 'count'
        ])
        print("\nCluster Statistics:")
        print(performance_stats)
        
    
        print("\n===== STATISTICAL SIGNIFICANCE TESTS =====")
        for metric in engagement_columns:
            try:
                cluster0_data = merged_data[merged_data['cluster'] == 0][metric]
                cluster1_data = merged_data[merged_data['cluster'] == 1][metric]
                
                if len(cluster0_data) == 0 or len(cluster1_data) == 0:
                    print(f"\nSkipping {metric} - insufficient data in clusters")
                    continue
                
              
                u_stat, p_value = stats.mannwhitneyu(cluster0_data, cluster1_data, 
                                                     alternative='two-sided')
                
                # Effect size calculation (r = Z / sqrt(N))
                n1, n2 = len(cluster0_data), len(cluster1_data)
                z_score = stats.norm.ppf(p_value/2)  # two-sided
                effect_size = abs(z_score) / np.sqrt(n1 + n2)
                
                print(f"\n{metric}:")
                print(f"p-value: {p_value:.4f} {'(Significant)' if p_value < 0.05 else '(Not significant)'}")
                print(f"Effect size: {effect_size:.4f}")
                print(f"Sample sizes: Cluster 0: {n1}, Cluster 1: {n2}")
                
            except Exception as e:
                print(f"Error analyzing {metric}: {str(e)}")

     
        plt.figure(figsize=(15, 5))
        for i, metric in enumerate(engagement_columns, 1):
            plt.subplot(1, 3, i)
            
            sns.boxplot(x='cluster', y=metric, data=merged_data)
            plt.yscale('log')
            plt.title(f'{metric} by Cluster')
            plt.xlabel('Cluster')
        plt.tight_layout()
        plt.show()
        
       
        print("\n===== AU-ENGAGEMENT CORRELATIONS =====")
        corr_data = []
        for col in au_columns:
            au_code = col.split('_')[0]
            au_name = au_meaning.get(au_code, "Unknown")
            for metric in engagement_columns:
                # Spearman correlation
                corr, p = stats.spearmanr(merged_data[col], merged_data[metric])
                corr_data.append({
                    'AU': col,
                    'AU_Name': f"{au_code}: {au_name}",
                    'Metric': metric,
                    'Correlation': corr,
                    'P_value': p,
                    'Significant': p < 0.05
                })
        corr_df = pd.DataFrame(corr_data)
       
        print("\nSignificant Positive Correlations:")
        significant_pos = corr_df[corr_df['Significant'] & (corr_df['Correlation'] > 0)].sort_values('Correlation', ascending=False)
        print(significant_pos[['AU_Name', 'Metric', 'Correlation', 'P_value']].head(10).to_string())
        print("\nSignificant Negative Correlations:")
        significant_neg = corr_df[corr_df['Significant'] & (corr_df['Correlation'] < 0)].sort_values('Correlation')
        print(significant_neg[['AU_Name', 'Metric', 'Correlation', 'P_value']].head(10).to_string())
      
        plt.figure(figsize=(12, 8))
        top_corrs = (
            corr_df[corr_df['Significant']]
            .sort_values('Correlation', key=abs, ascending=False)
            .head(15)
        )
        sns.barplot(x='Correlation', y='AU_Name', hue='Metric', data=top_corrs)
        plt.title('Top Facial Expression Correlations with Performance Metrics')
        plt.axvline(x=0, color='black', linestyle='-', alpha=0.3)
        plt.tight_layout()
        plt.show()
            
    except Exception as e:
        print(f"Error in analysis: {str(e)}")

analyze_cluster_performance(merged_data)
```

```python
def analyze_clusters_anova_raw(merged_data):
    """
    Perform ANOVA analysis for engagement metrics between clusters
    Using raw data with outlier removal
    """
    engagement_metrics = ['View Count', 'Like Count', 'Comment Count']
    results = []
    
    print("\n===== ANOVA Analysis of Clusters (Raw Data, No Outliers) =====")
    
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    for i, metric in enumerate(engagement_metrics):
       
        Q1 = merged_data[metric].quantile(0.25)
        Q3 = merged_data[metric].quantile(0.75)
        IQR = Q3 - Q1
        filtered_data = merged_data[
            (merged_data[metric] >= Q1 - 1.5 * IQR) & 
            (merged_data[metric] <= Q3 + 1.5 * IQR)
        ]
        
       
        cluster0_data = filtered_data[filtered_data['cluster'] == 0][metric]
        cluster1_data = filtered_data[filtered_data['cluster'] == 1][metric]
        
       
        f_stat, p_value = stats.f_oneway(cluster0_data, cluster1_data)
        
        
        metric_mean = filtered_data[metric].mean()
        ss_between = sum(len(group) * ((group.mean() - metric_mean) ** 2) 
                        for group in [cluster0_data, cluster1_data])
        ss_total = sum((filtered_data[metric] - metric_mean) ** 2)
        eta_squared = ss_between / ss_total
        
       
        results.append({
            'Metric': metric,
            'F_statistic': f_stat,
            'P_value': p_value,
            'Effect_size': eta_squared,
            'Significant': p_value < 0.05,
            'Mean_Cluster0': f"{cluster0_data.mean():,.0f}",
            'Mean_Cluster1': f"{cluster1_data.mean():,.0f}",
            'N_Cluster0': len(cluster0_data),
            'N_Cluster1': len(cluster1_data)
        })
        
      
        axes[i].grid(True, alpha=0.3)
        sns.boxplot(data=filtered_data, x='cluster', y=metric,
                   hue='cluster', ax=axes[i],
                   palette=['#2ecc71', '#e74c3c'],
                   legend=False)
        axes[i].set_title(f'{metric} Distribution\n(No Outliers)', pad=20)
        axes[i].set_xlabel('Cluster')
        axes[i].set_ylabel('Raw Count')
        
      
        axes[i].set_xticks([0, 1])
        axes[i].set_xticklabels(['Cluster 0', 'Cluster 1'])
        
        
        axes[i].yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: format(int(x), ',')))
        
    plt.suptitle('Distribution of Metrics Across Clusters (Outliers Removed)', 
                 y=1.02, fontsize=14)
    plt.tight_layout()
    plt.show()
    
    
    results_df = pd.DataFrame(results)
    
    print("\nDetailed ANOVA Results:")
    print("=" * 50)
    for _, row in results_df.iterrows():
        print(f"\n{row['Metric']}:")
        print(f"• F-statistic: {row['F_statistic']:.3f}")
        print(f"• P-value: {row['P_value']:.4f}")
        print(f"• Effect size (η²): {row['Effect_size']:.3f}")
        print(f"• Cluster 0 mean: {row['Mean_Cluster0']} (n={row['N_Cluster0']})")
        print(f"• Cluster 1 mean: {row['Mean_Cluster1']} (n={row['N_Cluster1']})")
        print(f"• Significant: {'Yes' if row['Significant'] else 'No'}")
    
    return results_df


anova_results_raw = analyze_clusters_anova_raw(merged_data)
anova_results_raw.to_csv('cluster_anova_results_raw.csv', index=False)
```
